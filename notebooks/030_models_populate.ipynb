```python
import sys

sys.path.append('../src')
```

```python
from domain.unit_of_work import SqlAlchemyUnitOfWork
from domain.models import Platform, Game, Tag ,Review, UserAccount, game_tag
```

```python
# 1. Creem la unitat de treball
uow = SqlAlchemyUnitOfWork()

# 2. Fem servir el bloc 'with' per gestionar automàticament la sessió
with uow:
    # --- A. BUIDAR DADES (Per començar de zero i evitar errors de duplicats) ---
    uow.session.execute(game_tag.delete())
    uow.session.query(Review).delete()
    uow.session.query(Tag).delete()
    uow.session.query(Game).delete()
    uow.session.query(Platform).delete()
    uow.session.query(UserAccount).delete()
    uow.commit() # Confirmem el buidat

    # --- B. CREAR PARES (Plataformes i Usuari) ---
    p1 = Platform(name="PC", manufacturer="Various")
    p2 = Platform(name="Nintendo Switch", manufacturer="Nintendo")
    uow.platforms.add(p1)
    uow.platforms.add(p2)

    u1 = UserAccount(username="Marti", email="marti@example.com", password="password123")
    uow.users.add(u1)
    
    # Fem commit aquí per forçar que la base de dades ens doni les IDs (p1.id, p2.id, u1.id)
    uow.commit() 

    # --- C. CREAR FILLS (Jocs, Tags i Ressenyes) ---
    g1 = Game(title="Elden Ring", price=59.99, platform_id=p1.id)
    g2 = Game(title="Mario Kart 8", price=49.99, platform_id=p2.id)
    uow.games.add(g1)
    uow.games.add(g2)

    t1 = Tag(tag_name="Souls-like")
    uow.tags.add(t1) # Fem servir el repositori de tags en lloc de session.add()

    r1 = Review(
        user_id=u1.id, 
        game_id=g1.id, 
        score=10, 
        comment="Una obra mestra absoluta!"
    )
    uow.reviews.add(r1)

    # --- D. GUARDAR TOT ---
    uow.commit()

print("Base de dades poblada amb èxit i sessió tancada correctament!")
```
